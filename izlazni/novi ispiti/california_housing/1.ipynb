### Na temelju znanja o navedenim algoritmima i na temelju svojstava promatranog podatkovnog skupa obrazložite koji od predloženih algoritama su prikladni za uˇcenje i predvi ¯danje cijene nekretnine na temelju ostalih znaˇcajki te koji su nedostatci ostalih gore navedenih algoritama. Predložite na koji naˇcin se dataset može prilagoditi za uˇcenje pojedinim algoritmom. 

1) linearna regresija
- prikladna za predviđanje cijene nekretnine jer imamo kontinuiranu izlaznu varijablu -> regresijski problem
- kvaliteta njegovog predviđanja ovisit će o linearnosti podataka

2) binarna logistička regresija
- nemamo klasifikacijski problem, tako da nam ovaj algoritam nije koristan
- mogli bismo ga npr. upotrijebiti kada bismo određivali samo je li kuća "skupa" ili "jeftina"

3) višeklasna klasifikacija logističkom regresijom
- ni ovaj algoritam nam nije koristan jer na izlazna varijabla nije kategorička
- kada bismo npr. određivali rangove cijena kuće, tada bismo mogli upotrijebiti

4) KNN
- može se koristiti i za regresiju, tako da bismo mogli upotrijebiti KNN algoritam
- kvaliteta njegovog predviđanja ovisit će o predobradi podataka (ponajviše skaliranje) te o odabranim hiperparametrima

5) SVM
- može se koristiti za regresiju pa ga možemo upotrijebiti
- kvaliteta njegovog predviđanja ovisit će ponajviše o odabiru hiperparametara

6) K-Means
- nije nam pogodan iz dva razloga: nije algoritam za nadzirano učenje, a mi imamo izlaznu varijablu, te je namijenjen za grupiranje podataka, a mi se bavimo regresijom
