Predložite na koji način se dataset može prilagoditi za učenje pojedinim algoritmom.

1) linearna regresija
- izlazna varijabla je kategorička stoga nam ovaj algoritam nije pogodan
- mogli bismo ga iskoristiti kada bismo npr. određivali postotak alkohola u vinu jer je on numerička varijabla

2) binarna logistička regresija
- radimo klasifikaciju, ali ne binarnu, pa nam ovaj algoritam nije koristan
- kada bismo imali samo dvije moguće vrijednosti za izlaznu varijablu mogli bismo ga koristiti

3) višeklasna klasifikacija logističkom regresijom
- ovaj algoritam bismo mogli iskoristiti za naš problem jer imamo višeklasnu klasifikaciju

4) KNN
- možemo iskoristiti ovaj algoritam jer se on može koristiti za višeklasnu klasifikaciju
- trebao bi dati dobre rezultate ako se dobro odabere broj susjeda

5) SVM
- može se koristiti za višeklasnu klasifikaciju pa ga možemo upotrijebiti
- kvaliteta njegovog predviđanja ovisit će ponajviše o odabiru hiperparametara

6) K-Means
- ovaj algoritam koristi se za nenadzirano učenje, a mi imamo izlaznu varijablu, pa nam nije idealan
- ipak koristi se za grupiranje podataka, pa iako vjerojatno nećemo dobiti baš one grupe koje odgovaraju našim klasama, bilo bi ga moguće iskoristiti
