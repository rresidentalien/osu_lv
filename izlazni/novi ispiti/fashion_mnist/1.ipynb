### Na temelju znanja o navedenim algoritmima i na temelju svojstava promatranog podatkovnog skupa obrazložite koji od predloženih algoritama su prikladni za uˇcenje i predvi ¯danje cijene nekretnine na temelju ostalih znaˇcajki te koji su nedostatci ostalih gore navedenih algoritama. Predložite na koji naˇcin se dataset može prilagoditi za uˇcenje pojedinim algoritmom. 

1) linearna regresija
- izlazna varijabla je kategorička stoga nam ovaj algoritam nije pogodan

2) binarna logistička regresija
- radimo klasifikaciju, ali ne binarnu, pa nam ovaj algoritam nije koristan

3) višeklasna klasifikacija logističkom regresijom
- ovaj algoritam bismo mogli iskoristiti za naš problem jer imamo višeklasnu klasifikaciju
- s obzirom da su podaci pikseli slike i time nisu linearni, upitno je hoće li dati dobar rezultat

4) KNN
- možemo iskoristiti ovaj algoritam jer se on može koristiti za višeklasnu klasifikaciju
- trebao bi dati dobre rezultate ako se dobro odabere broj susjeda

5) SVM
- može se koristiti za višeklasnu klasifikaciju pa ga možemo upotrijebiti
- kvaliteta njegovog predviđanja ovisit će ponajviše o odabiru hiperparametara

6) K-Means
- nije nam pogodan jer se mi bavimo nadziranim učenjem u ovom problemu, a k-means je algoritam nenadziranog učenja
- mogli bismo ga koristiti jer daje grupe, ali nije garantirano da će te grupe biti ekvivalentne našim klasama
